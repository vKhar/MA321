Write a program `is_palindrome()`  that:

- takes in the representation of a 9-digit combination code,
- returns the Boolean value `True` if the code is the same backwards and forwards, or
- returns the Boolean value `False` otherwise

<div style="text-align: right">[3]</div>

```python
## YOUR CODE HERE
def is_palindrome(code:str):
    return code[:4] == code[-1:-5:-1]
```

## Task 4

Write a program `is_different_consecutively()`  that:

- takes in the representation of a 9-digit combination code,
- returns the Boolean value `True` if all consecutive digits in the code are different,
- returns the Boolean value `False` otherwise

<div style="text-align: right">[3]</div>

```python
## YOUR CODE HERE
def is_different_consecutively(code:str):
    for i in range(len(code)-1):
        if code[i] == code[i+1]:
            return False
    else:
        return True
```

## Task 5

Write a program `low_digit_highest_freq()`  that:

- takes in the representation of a 9-digit combination code,
- returns the Boolean value `True` if digit with the lowest value in the code is used the highest number of times.
- returns the Boolean value `False` otherwise

<div style="text-align: right">[3]</div>

```python
## YOUR CODE HERE
def low_digit_highest_freq(code:str):

    low_digit = min( code)
    # digits_freq = {}
    # for digit in code:
    #     v = digits_freq.setdefault(digit, 0)
    #     digits_freq[digit] += 1
    digits_freq = { digit: code.count(digit) for digit in set(code)}
    low_digit_count = digits_freq[low_digit]
    # return digits_freq[low_digit] == max( digits_freq.values()) ## there may be other digits with the same max value eg 1234
    ## strictly greater than any other digit
    for d, count in digits_freq.items():
        if d != low_digit and count >=  low_digit_count:
            return False
    else:
        return True
```

## Task 6

Write a program `high_digit_lowest_freq()`  that:

- takes in the representation of a 9-digit combination code,
- returns the Boolean value `True` if digit with the highest value in the code is used the lowest number of times.
- returns the Boolean value `False` otherwise

<div style="text-align: right">[3]</div>

```python
## YOUR CODE HERE
def high_digit_lowest_freq(code:str):
    high_digit = max( code )
    digits_freq = { digit: code.count(digit) for digit in set(code) }
    high_digit_count = digits_freq[high_digit]

    for d, count in digits_freq.items():
        if d != high_digit and count <=  high_digit_count:
            return False
    else:
        return True
```

## Task 7

Write a program `solve_code()`  that:

- takes in no parameter
- returns the representation of the lock’s combination code based on the hints given in the question.

<div style="text-align: right">[2]</div>

```python
## YOUR CODE HERE
def solve_code():
    combinations = []
    for code in all_codes:
        if odd_last_digit(code) and \
        is_palindrome(code) and \
        is_different_consecutively(code) and \
        low_digit_highest_freq(code) and \
        high_digit_lowest_freq(code):
            combinations.append(code)
    return combinations
```

## Task 2
The organisation decided to rank the singers in each of the judge's lists according to the given scores as 1st, 2nd, 3rd, etc. For example, if a singer finished 4th, 2nd, 3rd and 6th on the judges' lists respectively, they would score a total of 4+2+3+6=15.

The winner of the contest is then determined by the smallest sum of all the rankings.

Write a program `winner()` that:
- implements the scoring criteria as described in the question
- returns a string that represents the winning singer for the above contest.

<div style="text-align: right">[5]</div>

##### Strategy:
- create a list winners to store ranking score for each singer
- for each singer, i
    - get the ranking(positions) from all judges. The sum gives the ranking score to determine the winner with the lowest score.
        - to get the ranking from each judge, j
            - put the scores of all singers awarded by that judge into a list. Sort the list in descending order
            - get the ranking positions for singer i from judge j
        - add the ranking scores for singer i from every judge into the wnners list
- finf the lowest score in the winners list, use the index of the lowest score to get the singer's name

```python
## YOUR CODE HERE
# if scores are the same they will have the lowest ranking [ 100,90,90,80,..], score with 90 will bith have a ranking of 2
def winner():
    winners = []
    for i in range(len(results)): # for each singer
        scores=0 # total scores based on ranking for all judges
        for j in range(len(results[0])): # number of judges
            score_from_judge = [ results[s][j] for s in range(len(results))] # scores for each singer from judge j
            ranked = sorted(score_from_judge, reverse=True)
            ranking = ranked.index( results[i][j]) + 1 # ranking from judge j
            # print(f"{names[i]} is ranked {ranking}")
            scores += ranking
        winners.append(scores)
    # print(winners)
    return names[winners.index(min(winners))] ## assume no same score

winner()
```

## Task 3
The organisation decided to host another competition with 26 competitors while keeping the same judges. However, due to the different popularity of the judges, they decided to give a weight to the judges scores.
- The scores that Jin gives to a singer will be multiplied with a factor of 0.3
- The scores that Hope gives to a singer will be multiplied with a factor of 0.2
- The scores that Rose gives to a singer will be multiplied with a factor of 0.4
- The scores that JungKook gives to a singer will be multiplied with a factor of 0.1

The new score table is given in the file `new_singers.csv` under the folder `resources` in the MA321 repo. Link: [new_singers.csv](https://github.com/beertino/MA321/blob/main/2025/Assignment/resources/new_singers.csv)

Write a program `new_winner()` that:
- takes in the nested list representation of the new score table,
- implements the new scoring criteria as described in Task 3,
- returns a string that represents the winning singer for the new contest.

<div style="text-align: right">[5]</div>

```python
## YOUR CODE HERE
from csv import reader
lines = list(reader(open("resources/new_singers.csv")))
names = [ line[0] for line in lines[1:]]
results = [ [int(s) for s in line[1:]] for line in lines[1:] ]

def new_winner():
    winners = []
    weights = [0.3, 0.2, 0.4, 0.1]
    for i in range(len(results)): # for each singer
        scores=0
        for j in range(len(results[0])): # number of judges
            score_from_judge = [ results[s][j]*weights[j] for s in range(len(results))]
            ranked = sorted(score_from_judge, reverse=True)
            ranking = ranked.index( results[i][j]*weights[j]) + 1
            # print(f"{names[i]} is ranked {ranking}")
            scores += ranking
        winners.append(scores)
    # print(winners)
    return names[winners.index(min(winners))] ## assume no same score


new_winner()
```

## Task 2

Write a program `beaver_message()`  that:

- takes in a letter combination consisting of a combination of letters of `PQRST`, and
- returns the representation, in the data type and value you specified, of the combination of flag positions.

<div style="text-align: right">[2]</div>

```python
## YOUR CODE HERE
def beaver_message(message):
    ret = []
    for letter in message:
        ret.append( codes[letters.index(letter)] )
    return "".join(ret)
```

## Task 4

Write a program `check_message()` that:

- takes in a letter combination and a combination of flag positions, and
- returns Boolean value `True` if the letter combination and the combination of flag positions matches and Boolean `False` otherwise.

<div style="text-align: right">[2]</div>

```python
## YOUR CODE HERE
def check_message(message, flag_positions):
    for letter in message:
        if letter == "P" and flag_positions[0] == "0":
            flag_positions = flag_positions[1:]
        elif letter == "Q" and flag_positions[0] == "1":
            flag_positions = flag_positions[1:]
        elif letter == "R" and flag_positions[:2] == "10":
            flag_positions = flag_positions[2:]
        elif letter == "S" and flag_positions[:3] == "001":
            flag_positions = flag_positions[3:]
        elif letter == "T" and flag_positions[:4] == "1001":
            flag_positions = flag_positions[4:]
        else:
            return False
    else:
        return True
```
